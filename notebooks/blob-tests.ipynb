```python
from azure.storage.blob import ContainerClient

# Replace with your Blob Storage account's container URL
container_url = "https://mldummyquaoutputsta.blob.core.windows.net/cv-inputs-blob"

# Create a ContainerClient for anonymous access
container_client = ContainerClient.from_container_url(container_url)

try:
    # List blobs in the container
    print("Listing blobs in the container:")
    blob_list = container_client.list_blobs()
    for blob in blob_list:
        print(f"Blob name: {blob.name}")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
from azure.storage.blob import BlobClient

# Replace with your Blob Storage SAS URL or full blob URL
blob_url = "https://mldummyquaoutputsta.blob.core.windows.net/cv-inputs-container"

# Replace with the local file path to the .txt file
local_file_path = "test.txt"

try:
    # Create a BlobClient
    blob_client = BlobClient.from_blob_url(blob_url)

    # Read the file and upload it
    with open(local_file_path, "rb") as data:
        blob_client.upload_blob(data, overwrite=True)
    print(f"File '{local_file_path}' uploaded successfully.")
except Exception as e:
    print(f"An error occurred: {e}")
```
