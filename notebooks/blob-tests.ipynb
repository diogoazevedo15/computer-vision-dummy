```python
from azure.storage.blob import ContainerClient

# Replace with your Blob Storage account's container URL
container_url = "https://mldummyquaoutputsta.blob.core.windows.net/cv-inputs-container"

# Create a ContainerClient for anonymous access
container_client = ContainerClient.from_container_url(container_url)

try:
    # List blobs in the container

    print("Listing blobs in the container:")
    blob_list = container_client.list_blobs()
    for blob in blob_list:
        print(f"Blob name: {blob.name}")
except Exception as e:
    print(f"An error occurred: {e}")
```

## Upload files

```python
from azure.storage.blob import BlobServiceClient

# Replace with your connection string (from the Azure portal)
connection_string = "<YOUR_CONNECTION_STRING>"

# Name of your container
container_name = "cv-inputs-container"

# Blob (file) name you want in the container
blob_name = "test.txt"

# Create the BlobServiceClient object
blob_service_client = BlobServiceClient.from_connection_string(connection_string)

# Get a client for the container
container_client = blob_service_client.get_container_client(container_name)

# Path to your local file
local_file_path = "test.txt"

try:
    with open(local_file_path, "rb") as data:
        # Upload the blob to the container
        container_client.upload_blob(
            name=blob_name, 
            data=data, 
            overwrite=True  # Set to True to overwrite if it already exists
        )
    
    print(f"Successfully uploaded '{local_file_path}' to the container '{container_name}' as '{blob_name}'.")
except Exception as ex:
    print(f"An error occurred while uploading: {ex}")
```
